```python
import torch
import PIL
from diffusers import  StableDiffusionPipeline, EulerAncestralDiscreteScheduler
from transformers import pipeline

model_id = "stable-diffusion-2-1-base"

pipeline = StableDiffusionPipeline.from_pretrained(
    model_id, 
)
pipeline.scheduler = EulerAncestralDiscreteScheduler.from_config(
    pipeline.scheduler.config
)

device = "cuda" if torch.cuda.is_available() else "cpu"
pipeline = pipeline.to(device)
```

```python
prompt = "A high-resolution product photo of modern running sneakers on a white background"

print("Generando imagen inicial del producto...")

imagen_producto = pipeline(
    prompt=prompt,
    num_inference_steps=30,    
    guidance_scale=7.5,        
    height=512,                
    width=512
).images[0]

nombre_archivo = "zapatillas_running_ecommerce.png"

imagen_producto.save(nombre_archivo)
prompts_variantes = {
    "estilo_flamenco": "A high-resolution product photo of flamenco style running sneakers",
    "estilo_urban": "A high-resolution product photo of urban style running sneakers",
    "estilo_bond": "A high-resolution product photo of James Bond style running sneakers",
}

for nombre_experimento, prompt_texto in prompts_variantes.items():
    print(f"Generando: {nombre_experimento}...")
    
    imagen_variante = pipeline(
        prompt=prompt_texto,
        num_inference_steps=30,
        guidance_scale=8.0,    
        height=512,
        width=512
    ).images[0]
```
